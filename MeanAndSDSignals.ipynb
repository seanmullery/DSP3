```python
import numpy as np
import matplotlib.pyplot as plt
ATU_navy =  "#001A79"
ATU_orange = "#FF791E"

plt.rcParams['axes.prop_cycle'] = plt.cycler(color=[ATU_navy])
plt.rcParams.update({
'text.color': ATU_navy,
'axes.labelcolor': ATU_navy,
'axes.titlecolor': ATU_navy,

'xtick.color': ATU_navy,
'ytick.color':ATU_navy,
})
```

```python


# User parameters
mean = 0.0      # Desired mean
sd = 1.0        # Desired standard deviation
N = 512         # Number of samples

# Generate random signal
signal = np.random.normal(loc=mean, scale=sd, size=N)

# Sample numbers
samples = np.arange(N)

# Plot
plt.figure(figsize=(12, 4))
plt.plot(samples, signal, linewidth=1)
plt.title(f'512-Point Random Signal (Mean={mean}, SD={sd})')
plt.xlabel('Sample Number')
plt.ylabel('Amplitude')
plt.grid(True, alpha=0.3)
plt.xlim(0, N-1)
plt.tight_layout()
plt.show()

# Display measured statistics
print(f"Measured Mean = {np.mean(signal):.4f}")
print(f"Measured SD   = {np.std(signal):.4f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

# Number of samples
N = 512
seed = 42

def plot_random_signal(mean=0.0, sd=1.0):
    """
    Generate and plot a 512-point random signal
    with the specified mean and standard deviation.
    """
    np.random.seed(seed)
    # Generate signal
    signal = np.random.normal(loc=mean, scale=sd, size=N)

    # Create plot
    plt.figure(figsize=(6, 4))
    plt.plot(signal, color=ATU_navy, linewidth=1)
    plt.title(f'512-Point Random Signal\nMean = {mean:.2f}, SD = {sd:.2f}')
    plt.xlabel('Sample Number')
    plt.ylabel('Amplitude')
    plt.ylim(-8, 8) # Fixed y-axis limits
    plt.xlim(0, N-1)
    plt.grid(True, alpha=0.3)
    plt.show()

    # Display actual statistics
    print(f"Measured Mean = {np.mean(signal):.4f}")
    print(f"Measured Standard Deviation = {np.std(signal):.4f}")

# Interactive controls
interact(
    plot_random_signal,
    mean=FloatSlider(
        value=0.0,
        min=-10.0,
        max=10.0,
        step=0.1,
        description='Mean'
    ),
    sd=FloatSlider(
        value=1.0,
        min=0.1,
        max=10.0,
        step=0.1,
        description='SD'
    )
)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

# Number of samples
N = 512
seed = 42

def plot_random_signal(mean=0.0, sd=1.0):
    """
    Generate and plot a 512-point random signal
    with the specified mean and standard deviation.
    """
    np.random.seed(seed)
    # Generate signal
    signal = np.random.normal(loc=mean, scale=sd, size=N)
    abs_signal = np.abs(signal-mean)
    
    # Create plot
    plt.figure(figsize=(6, 4))
    plt.plot(abs_signal, color=ATU_navy, linewidth=1)
    plt.title(f'512-Point Random Signal\nMean = {mean:.2f}, SD = {sd:.2f}')
    plt.xlabel('Sample Number')
    plt.ylabel('Amplitude')
    plt.ylim(-8, 8) # Fixed y-axis limits
    plt.xlim(0, N-1)
    plt.grid(True, alpha=0.3)
    plt.show()

    # Display actual statistics
    print(f"Measured Mean = {np.mean(abs_signal):.4f}")
    print(f"Measured Standard Deviation = {np.std(signal):.4f}")

# Interactive controls
interact(
    plot_random_signal,
    mean=FloatSlider(
        value=0.0,
        min=-10.0,
        max=10.0,
        step=0.1,
        description='Mean'
    ),
    sd=FloatSlider(
        value=1.0,
        min=0.1,
        max=10.0,
        step=0.1,
        description='SD'
    )
)
```
