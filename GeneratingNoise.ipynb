```python
import numpy as np
import matplotlib.pyplot as plt
# ==================================================
# ATU Colours
# ==================================================

ATU_navy = "#001A79"
ATU_orange = "#FF791E"
ATU_green = "#005B5E"
ATU_teal = "#7BB9CB"

plt.rcParams['axes.prop_cycle'] = plt.cycler(color=[ATU_navy])

plt.rcParams.update({
    'text.color': ATU_navy,
    'axes.labelcolor': ATU_navy,
    'axes.titlecolor': ATU_navy,
    'xtick.color': ATU_navy,
    'ytick.color': ATU_navy,
})

# -----------------------------
# Generate a random signal
# -----------------------------
N = 1000  # Number of samples

# Uniformly distributed random values between 0 and 1
x = np.random.uniform(0, 1, N)

# -----------------------------
# Plot results
# -----------------------------
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Time-domain signal
axes[0].plot(x, color=ATU_teal)
axes[0].set_title('Uniform Random Signal')
axes[0].set_xlabel('Sample Number')
axes[0].set_ylabel('Amplitude')
axes[0].grid(True)

# Histogram
axes[1].hist(x, bins=20, density=True, edgecolor='black')
axes[1].set_title('Histogram of Uniform Distribution')
axes[1].set_xlabel('Value')
axes[1].set_ylabel('Probability Density')
axes[1].grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider

# ==================================================
# ATU Colours
# ==================================================
ATU_navy = "#001A79"
ATU_orange = "#FF791E"
ATU_green = "#005B5E"
ATU_teal = "#7BB9CB"

plt.rcParams['axes.prop_cycle'] = plt.cycler(color=[ATU_navy])

plt.rcParams.update({
    'text.color': ATU_navy,
    'axes.labelcolor': ATU_navy,
    'axes.titlecolor': ATU_navy,
    'xtick.color': ATU_navy,
    'ytick.color': ATU_navy,
})

# ==================================================
# Interactive Plot
# ==================================================
def show_uniform_signal(
                        N,
                        num_signals):

    # Generate N uniformly distributed samples
    x = np.zeros(N)
    for i in range(num_signals):
        x = x+np.random.uniform(0, 1, N)

    fig, axes = plt.subplots(1, 2, figsize=(12, 4))

    # Time-domain signal
    axes[0].plot(x, color=ATU_teal)
    axes[0].set_title(f'Uniform Random Signal (N={N})')
    axes[0].set_xlabel('Sample Number')
    axes[0].set_ylabel('Amplitude')
    #axes[0].set_ylim(0, 1)
    axes[0].set_xlim(0,1000)
    axes[0].grid(True)

    # Histogram
    axes[1].hist(
        x,
        bins=100,
        density=False,
        color=ATU_orange,
        edgecolor='black'
    )
    axes[1].set_title('Histogram of Uniform Distribution')
    axes[1].set_xlabel('Value')
    axes[1].set_ylabel('Count')
    #axes[1].set_xlim(0, 1)
    axes[1].grid(True)

    plt.tight_layout()
    plt.show()

    print(f"Mean is {np.mean(x)}")
    print(f"StdDev is {np.std(x)}")

# Slider
interact(
    show_uniform_signal,
    N=IntSlider(
        value=100,
        min=10,
        max=1000000,
        step=10,
        description='N'
    ),
     num_signals=IntSlider(
        value=1,
        min=1,
        max=20,
        step=1,
        description='Num signals'
    )
    
);
```
