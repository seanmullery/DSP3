```python
import numpy as np
import matplotlib.pyplot as plt
import time

from ipywidgets import interact, IntSlider, FloatSlider, Checkbox

# ==================================================
# ATU Colours
# ==================================================

ATU_navy = "#001A79"
ATU_orange = "#FF791E"
ATU_green = "#005B5E"
ATU_teal = "#7BB9CB"

plt.rcParams['axes.prop_cycle'] = plt.cycler(color=[ATU_navy])

plt.rcParams.update({
    'text.color': ATU_navy,
    'axes.labelcolor': ATU_navy,
    'axes.titlecolor': ATU_navy,
    'xtick.color': ATU_navy,
    'ytick.color': ATU_navy,
})

# ==================================================
# Main Function
# ==================================================

def show_signal_histogram(mean=128,
                          std_dev=30,
                          N=1000,
                          normalized=False):

    # --------------------------------------------------
    # Generate Signal
    # --------------------------------------------------

    np.random.seed(42)

    signal = np.random.normal(mean, std_dev, N)

    signal_clipped = np.clip(signal, 0, 255)

    signal_q = np.round(signal_clipped).astype(np.uint8)

    # ==================================================
    # METHOD 1
    # DIRECT CALCULATION
    # ==================================================

    t0 = time.perf_counter()

    sum_x = 0
    
    for sample in signal_q:
        sum_x += int(sample)
    
    mean_direct = sum_x / len(signal_q)
    
    sum_var = 0
    
    for sample in signal_q:
        diff = sample - mean_direct
        sum_var += diff * diff
    
    std_direct = np.sqrt(sum_var / len(signal_q))
    
    direct_time = (time.perf_counter() - t0) * 1000

    # ==================================================
    # METHOD 2
    # HISTOGRAM CALCULATION
    # ==================================================

    # ==================================================
    # Histogram Construction
    # ==================================================
    t0 = time.perf_counter()

    histogram = [0] * 256
    
    for sample in signal_q:
        histogram[int(sample)] += 1
    
    hist_build_time = (time.perf_counter() - t0) * 1000

    t0 = time.perf_counter()

    # ==================================================
    # Statistics via Histogram 
    # ==================================================
    total = 0
    
    for i in range(256):
        total += i * histogram[i]
    
    mean_hist = total / len(signal_q)
    
    variance = 0
    
    for i in range(256):
        diff = i - mean_hist
        variance += histogram[i] * diff * diff
    
    std_hist = np.sqrt(
        variance / len(signal_q)
    )
    
    hist_stats_time = (
        time.perf_counter() - t0
    ) * 1000

    # ==================================================
    # OPERATION COUNT ESTIMATES
    # ==================================================

    direct_ops = 2 * N

    histogram_ops = N + 512

    # ==================================================
    # PLOTS
    # ==================================================

    fig, axs = plt.subplots(1, 3, figsize=(18, 5))

    # --------------------------------------------------
    # Signal Plot
    # --------------------------------------------------

    display_samples = min(1000, len(signal_q))

    axs[0].plot(
        signal_q[:display_samples],
        color=ATU_teal
    )

    axs[0].set_title(
        f"Quantized Signal\n(First {display_samples} Samples)"
    )

    axs[0].set_xlabel("Sample Number")

    axs[0].set_ylabel("Amplitude")

    axs[0].set_ylim([0, 255])

    axs[0].grid(True)

    # --------------------------------------------------
    # Histogram
    # --------------------------------------------------

    axs[1].hist(
        signal_q,
        bins=np.arange(257) - 0.5,
        density=normalized,
        color=ATU_orange,
        edgecolor=ATU_orange
    )

    axs[1].set_title("Histogram")

    axs[1].set_xlabel("Value")

    if normalized:
        axs[1].set_ylabel("Probability Density")
        axs[1].set_title("Normalised Histogram")
    else:
        axs[1].set_ylabel("Count")
        axs[1].set_title("Histogram")

    axs[1].set_xlim([0, 255])

    axs[1].grid(True)

    # --------------------------------------------------
    # Computational Cost
    # --------------------------------------------------

    # --------------------------------------------------
    # Computational Cost (Stacked Bar)
    # --------------------------------------------------

    axs[2].bar(
        ["Direct Method"],
        [direct_time],
        color=ATU_teal,
        label="Mean + SD from samples"
    )
    
    axs[2].bar(
        ["Histogram Method"],
        [hist_build_time],
        color=ATU_orange,
        label="Build Histogram"
    )
    
    axs[2].bar(
        ["Histogram Method"],
        [hist_stats_time],
        bottom=[hist_build_time],
        color=ATU_green,
        label="Mean + SD from Histogram"
    )

    axs[2].text(
        0,
        direct_time,
        f"{direct_time:.3f} ms",
        ha='center',
        va='bottom'
    )
    
    axs[2].text(
        1,
        hist_build_time + hist_stats_time,
        f"{hist_build_time + hist_stats_time:.3f} ms",
        ha='center',
        va='bottom'
)

    axs[2].set_ylabel("Execution Time (ms)")
    
    axs[2].set_title("Measured Execution Time")
    
    axs[2].grid(True, axis='y')
    
    axs[2].legend()

    plt.show()

    # ==================================================
    # RESULTS
    # ==================================================

    print()
    print("=" * 65)
    print("DIRECT CALCULATION FROM SAMPLES")
    print("=" * 65)

    print(f"Mean               = {mean_direct:.3f}")
    print(f"Standard Deviation = {std_direct:.3f}")
    print(f"Execution Time     = {direct_time:.4f} ms")

    print()
    print("=" * 65)
    print("CALCULATION FROM HISTOGRAM")
    print("=" * 65)

    print(f"Mean               = {mean_hist:.3f}")
    print(f"Standard Deviation = {std_hist:.3f}")
    print(f"Histogram Build    = {hist_build_time:.4f} ms")
    print(f"Histogram Stats    = {hist_stats_time:.4f} ms")
    print(
        f"Total Time         = "
        f"{hist_build_time + hist_stats_time:.4f} ms"
    )
    print(f"Number of samples with value 100: {histogram[100]}")
    print(f"Number of samples with value 150: {histogram[150]}")
    print(f"Number of samples with value 200: {histogram[200]}")
    
    
# ==================================================
# INTERACTIVE CONTROLS
# ==================================================

interact(
    show_signal_histogram,

    mean=IntSlider(
        value=128,
        min=-100,
        max=355,
        step=1,
        description='Mean'
    ),

    std_dev=FloatSlider(
        value=30,
        min=1,
        max=150,
        step=1,
        description='Std Dev'
    ),

    N=IntSlider(
        value=1000,
        min=10,
        max=500000,
        step=100,
        description='Length'
    ),
    
    normalized=Checkbox(
        value=False,
        description='Normalise Histogram'
    )
    
);
```
